## See a problem that can be solved using python and math module.
Fruit-fly problem

Problem statement: The colony size of fruit-flies (IU is famous for mainining a large stock of different flies) is given by this formula: $P(t) = P_0 e^{kt}$ where $P_0$ is the intial size of the colony, and $P(t)$ is the colony size after time $t$. Assume a colony doubles in size in 12 days. The question is, given initial size of 1000, what's the size of the colony in 10 days?

Analysis: to compute the colony size, one needs to know the $k$, which can be determined using the given formula and assumption (a colony doubles in 12 days). To compute $k$, we need to transform the given formula: $k = log(P(t)/P(0))/t$, and we know $P(t) / P(0)$ is 2 (the colony doubles) when $t=12$. BTW, math.log by default uses $e$ as the base.   

````python
import math
t = 12
k = math.log(2) / 12

P0, t = 1000, 10
Pt = P0 * math.e ** (k * t)

print(f"k = {k}, Pt = {Pt}")
print(f"After {t} days, the fruit-fly colony of size {P0} grows into size of {int(Pt)}")
````

# Choice (branching structures)

Here we use a toy example: car insurance premium as an example to demonstrate the choice.

````python
def premium(age):
  high, low = 1500, 400
  young, mature = 20, 35
  if age <= young:
    return high
  elif age < mature:
    slope = (low - high) / (mature - young)
    return int(high + slope * (age - young))
  else:
    return low

for age in [19, 23, 40, 50]:
  print(f"age {age} premium {premium(age)}")
````

If you are the type of programmer who wants to compress your codes, you may want to having if and else statements in one line as shown below. The functions are to computes the sum of even numbers in a given list of numbers.  

````python
def add_even_1(nums):
  """compressed way"""
  add = 0
  for n in nums:
    add += n if n % 2 == 0 else 0
  return add

def add_even_2(nums):
  """uncompressed way"""
  add = 0
  for n in nums:
    if n % 2 == 0:
      add += n
    else: #this else statement is not needed, kept here for demonstration purpose
      add += 0
  return add

nums = [2, 30, 5, 9, 1111, 20]
print(f"the sum of even nums in {nums} is {add_even_1(nums)}, {add_even_2(nums)}")
````

# Loops

* there are two types of loops, while loop (unbounded) and for loop (bounded).
* here we use a simple problem to demonstrate the usage of the loops
* the problem: given n, compute the sum from 1, 2, all the way to n.

````python
def sum_n_for(n):
  add = 0
  for i in range(1, n + 1):
    add += i
  return add


def sum_n_while(n):
  add = 0
  i = 0
  while i <= n:
    add += i
    i += 1
  return add


n = 100
print(f"1 + 2 + .. + {n} = {sum_n_for(n)}, {sum_n_while(n)}")
````

Iterable types (i.e., containers) are designed to work with bounded loops.
There are two ways: used directly as iterable, or indirectly by using subscript.

````python
nums = [10, 40, 50, 70, 90]

#using the list as iterable directly
sum1 = 0
for anum in nums: #loop through the iterable
  sum1 += anum

#using the list indirectly by subscripting
sum2 = 0
for i in range(len(nums)): #loop through the indices
  sum2 += nums[i]

print(f"sum of {nums} is {sum1}, {sum2}")
````

Break and continue are two statements that may be used to work with a loop to change its behavior.

````python
add = 0
for i in [10, 29, 40, 46, 50, 100, 300]:
  if i % 5 == 0:
    continue
  if i > 100:
    break
  add += i

print(add)

````

Here we use a program that approximates the number $e$ (Euler's number) (the base of natural logarithms). It is approximately equal to 2.71828. It is the limit of $(1 + 1/n)^n$ as n approaches infinity. It can also be calcualted as the sum of the infinite series,
$e = \sum_{n=0}^{\infty} \frac{1}{n!} = 1 + \frac{1}{1} + \frac{1}{1 × 2} + \frac{1}{1 × 2 × 3} + \dots $.

````python
import math
def approx_e(error):
  e_0, e_1, i = 1, 2, 2
  while abs(e_0 - e_1) > error:
    e_0, e_1, i = e_1, e_1 + 1/math.factorial(i), i + 1
  return e_1

print(f"Approximations of e with different tolerance: {approx_e(0.0001)} {approx_e(1/10000000)}")
````

# Recursion

To use recursion, one needs to figure out the base case, and the induction. Below we use a simple example, to compute the sum for a given list using a recursion.

````python
def rec_sum(nums):
  if nums == []:
    #base case, return the solution directly
    return 0
  else:
    #induction, the solution relies on calling the function itself on a smaller problem
    #(for this case, the sum of a one-element smaller list of numbers)
    return nums[0] + rec_sum(nums[1:])

nums = [10, 29, 50, 60]
print(f"sum of {nums} is {rec_sum(nums)}")
````

Factorial is another good example to demonstrate recursion.
```
f(0) = 1
f(n) = n * f(n - 1)
```

````python
def fac(n):
  """compute factorial of n using recursion"""
  if n == 0:
    #base
    return 1
  else:
    #induction
    return n * fac(n - 1)
print(f"factorial of {10} is {fac(10)}")
````

# Object-oritented programming (OOP)

Class -- the blueprint. Classes provide a means of bundling data (attributes) and functionality (methods) together. 
Object -- the instances. Objects can be created from classes. 
Below we use a simple class Pet as the toy example. Once we create the class, we can create many pets (instances) from the class. 

````python
class Pet:
    """This is the Pet class"""
    def __init__(self, name):
        self.name = name
    def set_name(self, foo):
        self.name = foo
    def get_name(self):
        return self.name
    def talk(self):
        print("bork, bork!")
print(Pet.__doc__) #print the class's doc string
p1 = Pet("Ritz") #create an instance named p1, pass "Ritz" to p1.name
p1.talk() #note, p1.talk(), not Pet.talk()
p2 = Pet("Max")
print(f"I am {p2.get_name()}")
p2.talk()
````

# Search problem and algorithms (e.g., binary search)

````python
def linearsearch(X, y):
  for i in X:
    if y == i:
      return 1
  return 0

def binarysearch(X, y):
  if not X:
    return 0
  else:
    mid = int(len(X) / 2)
    if X[mid] == y:
      return 1
    elif X[mid] > y:
      return binarysearch(X[:mid], y)
    else:
      return binarysearch(X[mid+1:], y)
data = [2, 7, 8, 10, 26, 70, 80, 100]
query = [2, 8, 90]
for q in query:
  print(f"{q}: {linearsearch(data, q)} {binarysearch(data, q)}")
````
