### Why Linux/Unix?
- Bioinformatics has historically been built around the Unix/Linux command-line environment. Many widely used tools, e.g., BLAST, BWA, minimap2, SAMtools, BEDTools, HMMER, are designed primarily for Linux/Unix.
- Linux makes it easy to chain tools together using pipes:
  e.g, minimap2 reference.fa reads.fastq | samtools sort -o aligned.bam
- Most HPC clusters and cloud computing environments run Linux, so Linux skills transfer directly to real research environments.
- Package/environment managers such as Conda/Bioconda make installing complex bioinformatics software much easier.

### Quick Biology Primer

- **DNA (Deoxyribonucleic Acid)**: Contains genetic instructions using 4 nucleotides: Adenine (A), Thymine (T), Guanine (G), Cytosine (C)
- **Transcription**: DNA → RNA (T becomes U)
- **Translation**: RNA → Protein (groups of 3 nucleotides = 1 amino acid)
- **Codon**: A group of 3 nucleotides that codes for one amino acid
- **GC Content**: The percentage of G and C nucleotides in a sequence (affects DNA stability)

Biology provides excellent real-world examples for programming:
- **DNA sequences**: Long strings of A, T, G, C nucleotides
- **Protein sequences**: Chains of 20 different amino acids
- **Structured problems**: Clear rules for transformation and analysis
- **Practical applications**: Detect diseases, find mutations, design drugs

## Section 7: Basic File I/O with FASTA Format

### What is FASTA?

FASTA is the standard format for storing biological sequences:

```
>Sequence_ID [description]
ATGGTCTACCTAGATCGCCGC

```

Key features:
- Lines starting with `>` are headers (contains ID and metadata)
- Following lines contain the sequence
- Sequences can span multiple lines
- Used by NCBI, UniProt, and all major databases

### Tips for Success
- **Run code cells repeatedly** - Modify values and see what changes
- **Experiment fearlessly** - Python won't break your computer!
- **Read error messages** - They often tell you exactly what's wrong
- **Write your own functions** - Don't just copy-paste
- **Document your code** - Explain what each function does
- **Use meaningful variable names** - `gc_percent` is better than `gp`

### Troubleshooting Common Issues

**IndexError: string index out of range**
- You tried to access a position that doesn't exist
- Check your loop ranges and slicing indices

**TypeError: 'str' object is not subscriptable**
- You're trying to index something that isn't a string or list
- Verify the variable type using `type(variable)`

**FileNotFoundError**
- The file path is incorrect
- Check that the file exists in the location you specified
- Use absolute paths if relative paths don't work

**UnboundLocalError**
- You used a variable before defining it
- Make sure variables are defined before the code that uses them
